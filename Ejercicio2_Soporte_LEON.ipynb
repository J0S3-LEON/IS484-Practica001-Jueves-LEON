## <span style="color:#1f497d; border-bottom: 2px solid #1f497d; padding-bottom: 5px;">Ejercicio 2: Escalamiento de tickets de soporte técnico</span>

### <span style="color:#c0504d;">Ficha PEAS</span>

* <b style="color:#4f81bd;">Percepción (S):</b> Tiempo de espera en minutos (entero), nivel de urgencia (texto), cliente premium (booleano).
* <b style="color:#4f81bd;">Acciones (A):</b> Escalar a Nivel 1, Escalar a Nivel 2, Escalar a Nivel 3.
* <b style="color:#4f81bd;">Entorno (E):</b> Sistema de mesa de ayuda (Help Desk).
* <b style="color:#4f81bd;">Objetivo:</b> Resolver problemas priorizando a clientes importantes y urgencias, respetando el SLA (Acuerdos de Nivel de Servicio).
* <b style="color:#4f81bd;">Medida de desempeño (P):</b> Tiempo promedio de resolución, tasa de satisfacción del cliente, cumplimiento de SLA.

### <span style="color:#c0504d;">Justificación de las reglas</span>

<span style="color:#595959;">Las reglas aseguran la eficiencia y lealtad del cliente. Un cliente "premium" o una urgencia "alta" van directo al Nivel 3 para atención especializada e inmediata. Si la urgencia es "media" y la espera supera los 30 minutos, o si es "baja" y supera los 60 minutos, se escala al Nivel 2 preventivamente para evitar romper los acuerdos de servicio (SLA). Los demás casos regulares (esperas cortas de urgencia media o baja) se inician en el Nivel 1.</span>
